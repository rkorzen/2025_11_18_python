```python
### napisz uniwersalny filtr, ktory wybierz z zadanej listy elementy pasujace do podanej funkcji

# def filtr(collection, key):
#     output = []
#     for c in collection:
#         if key(c):
#             output.append(c)
#     return output


def filtr(collection, key):
    return [c for c in collection if key(c)]

assert filtr([1, 2, 3], key=lambda x: x % 2 == 0) == [2]
assert filtr(["Ania", "Anatol", "Gienia"], key=lambda x: x.startswith("A")) == ["Ania", "Anatol"]
```
