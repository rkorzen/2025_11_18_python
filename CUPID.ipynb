# Composition over inheritance



```python

# dziedziczenie
class Animal:
    sound = ""

    def voice(): return self.sound

class Dog(Animal):
    sound = "Woof"
    def voice(): return "Woof"

class Cat(Animal):
    sound = "Miau"
    def voice(): return "Woof"



# kompozycja

class Speaker:
    sound = ""

    def voice(self): return self.sound

class Woof(Speaker):
    sound = "Woof"


class Miau(Speaker):
    sound = "Miau"


class Animal:

    def __init__(self, speaker: Speaker):
        self.speaker = speaker()

    def voice(self):
        return self.speaker.voice()

a = Animal(Woof)
a.voice()
```

U – Unix philosophy (zasada filozofii Unix)

Program/komponent powinien robić jedną rzecz i robić ją dobrze.
To bardziej praktyczne podejście niż „single responsibility”, bo odnosi się do faktycznej funkcjonalnej roli, a nie abstrakcji.

P – Predictable (przewidywalny)

Kod powinien zachowywać się w sposób intuicyjny, bez zaskoczeń.
Ta sama funkcja/rzecz powinna dawać ten sam efekt, zawsze w tych samych warunkach.

I – Idiomatic (idiomatyczny)

Kod powinien być zgodny z konwencjami danego języka.
W Pythonie: używamy with, context managerów, list compów, dataclass, enumerate, itd.
Czyli: nie walcz z językiem — korzystaj z jego naturalnych wzorców.

D – Domain-based (oparty na domenie)

Kod powinien odzwierciedlać pojęcia domenowe, a nie techniczne detale.
Klasy, metody i moduły mają mówić językiem biznesu/domeny, a nie frameworka lub bazy danych.
