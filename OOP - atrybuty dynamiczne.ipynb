```python
class Square:

    def __init__(self, side):

        self.side = side

    @property
    def side(self):
        return self._side

    @side.setter
    def side(self, value):
        if value <= 0:
            raise ValueError("Side cannot be negative")
        self._side = value

    
    @property # atrybit wyliczalny, dynamiczny
    def rectangle(self):
        return self.side * 4

    @rectangle.setter
    def rectangle(self, value):
        self.side = value / 4

    # zrob to samo dla area

    @property # atrybit wyliczalny, dynamiczny
    def area(self):
        return self.side ** 2

    @area.setter
    def area(self, value):
        self.side = value ** 0.5


    def __add__(self, other):
        area = self.area + other.area
        return self.__class__(area**0.5)
        

    def __repr__(self):
        return f"Square({self.side})"

    # def __str__(self):
    #     return f"xxxxSquare({self.side})"

    def __eq__(self, other):
        return self.side == other.side

    def __lt__(self, other):
        return self.side < other.side

    def __ge__(self, other):
        return self.side >= other.side


s = Square(10)
s2 = Square(10)
s.area
print(s + s2)
print([s + s2])
```
