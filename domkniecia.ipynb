```python
def incr_factory(step):

    def incr(x):
        return x + step

    return incr
```

```python
lista = [1, 2, 4, 5, 7, 20]
```

```python
liczba = 10
```

```python
for step in lista:
    func = incr_factory(step)
    print(func(liczba))
```

```python
### napisz funkcje make_censor(bad_words)

bad_words = ["kot", "burak"]

def make_censor(bad_words):

    def add_word(word):
        bad_words.append(word)
    
    def censor(text):
        words = text.split()
        output = []
        for word in words:
            if word.lower() in bad_words:
                output.append("*" * len(word))
            else:
                output.append(word)
        return " ".join(output)
    
    return censor, add_word

censor, add_stop_word_to_censor = make_censor(bad_words)
censor2, add_stop_word_to_censor2 = make_censor(["moj", "je"])

assert censor("Moj kot je burak") == "Moj *** je *****"
assert censor("Moj kot je burak") == "Moj *** je *****"
assert censor("Moj KOT je BURAK") == "Moj *** je *****"


assert censor2("Moj kot je burak") == "*** kot ** burak"

add_stop_word_to_censor2("kot")

assert censor2("Moj kot je burak") == "*** *** ** burak"
```
