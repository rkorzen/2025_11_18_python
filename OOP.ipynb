```python
# Klasy

class Person:

    def __init__(self, name, b_year):
        self.name = name
        self.b_year = b_year


    def hello(self):
        return f"Hi I'am {self.name}"

    def age(self):
        return 2025 - self.b_year


class Worker(Person):

    def __init__(self, name, b_year, year_of_start_work):
        super().__init__(name, b_year)
    
        if year_of_start_work - b_year < 18:
            raise ValueError("Nie powinien zaczac pracy przed ukonczeniem 18 r zycia")
            
        self.year_of_start_work = year_of_start_work


    def staz(self):
        return 2025 - self.year_of_start_work
        
osoby = [
    Person("Rafał", 1980),  
    Person("Bronek", 1956),
    Worker("Cenia", 2002, 2022)
]

for o in osoby:
    print(o.hello())
    print(o.age())
    if hasattr(o, "staz"):
        print(o.staz())
```

```python
# Funkcje

def hello(name):
    return f"Hi I'am {name}"

def age(year):
    return 2025 - year

def staz(osoba):
    if len(osoba) == 3:
        return 2025 - osoba[2]
    return None

osoby = [
    ["Rafal", 1980],
    ["Bronek", 1956],
    ["Cenia", 2002, 2022]
    
]

for osoba in osoby:
    print(hello(osoba[0]))
    print(age(osoba[1]))
    if staz(osoba):
        print(staz(osoba))
```

```python

class Person:

    def __init__(self, name, b_year):
        self.name = name
        self.b_year = b_year


class Worker(Person):

    def __init__(self, name, b_year, year_of_start_work):
        super().__init__(name, b_year)
    
        if year_of_start_work - b_year < 18:
            raise ValueError("Nie powinien zaczac pracy przed ukonczeniem 18 r zycia")
            
        self.year_of_start_work = year_of_start_work



def hello(p: Person):
    return f"Hi I'am {p.name}"

def age(p: Person):
    return 2025 - p.year

def staz(p: Worker):
    if not isinstance(p, Worker):
        return None
    return 2025 - p.year_of_start_work
    
```
