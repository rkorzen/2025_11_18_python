```python
class Bonus:
    instance_count = 0

    def __init__(self):
        Bonus.instance_count += 1

    def __del__(self):
        Bonus.instance_count -= 1
        
```

```python
a, b, c = Bonus(), Bonus(), Bonus()
```

```python
class Kwadrat:
    def __init__(self, bok):
        self.bok = bok

    
    def pole_powierzchni(self): # metoda instnacji - pierwszy argument to instancja - na ogol nazywamy ja self
        return self.bok ** 2

    @classmethod  # metoda klasowa
    def tworz(cls, num): # pierwszy argument to klasa - w konwencji nazwa cls
        output = []
        for i in range(num):
            # output.append(Kwadrat(1))
            output.append(cls(1))
        return output
            
    @staticmethod
    def tworz2(num):
        output = []
        for i in range(num):
            # output.append(Kwadrat(1))
            output.append(Kwadrat(1))
        return output


    
```

```python
kw = Kwadrat.tworz(5)
```

```python
kw[0].tworz(10)
```
