# **“Znajdź Skarb”**

Twoim zadaniem jest zaimplementować prostą, konsolową grę logiczną pod tytułem **„Znajdź Skarb”**. Gracz porusza się po dwuwymiarowej planszy, a celem gry jest odnalezienie ukrytego skarbu w jak najmniejszej liczbie ruchów.

## **Wymagania funkcjonalne**

1. **Plansza**

   * Gra odbywa się na prostokątnej planszy o zadanej szerokości i wysokości.
   * Pozycje na planszy reprezentowane są współrzędnymi `(x, y)`, gdzie `(0, 0)` to lewy dolny róg.

2. **Gracz**

   * Gracz startuje w losowym miejscu na planszy.
   * Może poruszać się o jedno pole w czterech kierunkach:

     * **w** – góra
     * **s** – dół
     * **a** – lewo
     * **d** – prawo
   * Ruch poza planszę jest zabroniony — należy zgłosić błąd.

3. **Skarb**

   * Skarb umieszczany jest losowo na planszy, ale **nie może znajdować się na pozycji startowej gracza**.
   * Po wejściu gracza na pole skarbu gra powinna się zakończyć komunikatem o zwycięstwie i liczbie wykonanych ruchów.

4. **System podpowiedzi (ciepło / zimno)**

   * Po każdym ruchu program oblicza dystans Manhattanowski między graczem a skarbem.
   * Gracz otrzymuje komunikat:

     * **„Ciepło”** – jeśli przybliżył się do skarbu,
     * **„Zimno”** – jeśli się oddalił,
     * **„Bez zmian”** – jeśli dystans pozostał taki sam.

5. **Interfejs konsolowy**

   * Po każdym ruchu należy wyświetlić planszę:

     * `P` – pozycja gracza
     * `T` – pozycja skarbu *(tylko w trybie debug)*
     * `.` – puste pole
   * Program powinien działać w pętli do momentu znalezienia skarbu.

6. **Architektura**

   * Program powinien być podzielony na logiczne komponenty:

     * `Board` – plansza gry
     * `Position` – pojedyncza pozycja
     * `Player` – gracz
     * `Treasure` – skarb
     * `Game` – logika gry
     * `GameView` + `ConsoleView` – interfejs wyświetlania
     * `GameController` – pętla sterująca całym przebiegiem gry
   * Obiekty powinny być zaprojektowane obiektowo i w sposób czytelny.

7. **Tryb debug**

   * Po jego włączeniu program powinien:

     * pokazywać pozycję skarbu,
     * oznaczać ją na planszy literą `T`.

## **Cel zadania**

Stwórz pełną implementację gry, uwzględniając powyższe zasady. Program powinien:

* losować pozycje startowe,
* poprawnie walidować ruchy,
* wyświetlać podpowiedzi ciepło/zimno,
* wyświetlać aktualny stan planszy,
* zakończyć się po odnalezieniu skarbu.

```python
# %%writefile game_v1.py
from random import randint

class Position:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def move(self, dx, dy):
        return Position(self.x + dx, self.y + dy)

    def distance(self, other):
        return abs(self.x - other.x) + abs(self.y - other.y)
    
    def __repr__(self):
        return f"({self.x}, {self.y})"

    def __eq__(self, other):
        return self.x == other.x and self.y == other.y
        

MOVES = {
    "w": (0, 1),
    "s": (0, -1),
    "a": (-1, 0),
    "d": (1, 0)
}



def play(width=10, height=10, debug=False):
    print("Start. Znajdz Skarb")

    player = Position(randint(0, width-1), randint(0, height-1))
    treasure = Position(randint(0, width-1), randint(0, height-1))

    # Skarb nie moze byc na tej samej pozycji
    while player == treasure:
        treasure = Position(randint(0, width-1), randint(0, height-1))

    print(f"Twoja pozycja startowa to: {player}")

    if debug:
        print(f"[DEBUG] Skarb: {player}")


    last_distance = player.distance(treasure)
    moves = 0

    while True:
        cmd = input("Podaj kierunek [w a s d]").strip().lower()

        if cmd == "q":
            break
            
        if cmd not in MOVES:
            print("Niepoprawny kierunek")
            continue

        dx, dy = MOVES[cmd]

        new_pos = player.move(dx, dy)

        if new_pos.x < 0 or new_pos.x > width - 1 or new_pos.y < 0 or new_pos.y > height -1:
            print("trafiles na sciane")
            continue
            
        print("Twoja pozycja:", new_pos)
        player = new_pos

        if player == treasure:
            print("Brawo! Znalazles Skarb")
            break

        dist = player.distance(treasure)

        if dist < last_distance:
            print("Cieplo")
        else:
            print("Zimno")

        last_distance = dist

play()
```
