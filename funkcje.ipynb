````python
%%writefile xxx.py
def formatuj(*args, **kwargs):
    text = "\n\n".join(args)
    for k, v in kwargs.items():
        text = text.replace(f"${k}", str(v))
    return text


assert formatuj() == ""
assert formatuj("ala ma kota") == "ala ma kota"
assert formatuj("a", "b") == "a\nb"
assert formatuj("a", "b", imie="Jacek") == "a\nb"
assert formatuj("a $imie $xxx", "b $xxx", imie="Jacek", xxx="cos tam") == "a Jacek cos tam\nb cos tam"
assert formatuj("cena: $cena", "vat: $vat", cena="10", vat="23") == "cena: 10\nvat: 23"


````

## rekurencja

````python
def silnia(n):
    if n == 0:
        return 1
    return n * silnia(n-1)
````

````python
silnia(10)
````

````python
def splaszcz(lista: list[int|list]) -> list[int]:
    output: list[int] = []
    for el in lista:
        if type(el) is list:
            output.extend(splaszcz(el))
        else:
            output.append(el)
    return output

assert splaszcz([]) == []
assert splaszcz([1, 2, 3]) == [1, 2, 3]
assert splaszcz([1, 2, [3]]) == [1, 2, 3]
assert splaszcz([1, 2, [3, 4, [5, 6]]]) == [1, 2, 3, 4, 5, 6]
````
