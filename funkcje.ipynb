```python
def funkcja(a, b, c=1, d=2):  # snake_case
    print(a, b, c, d)

# poslugiwanie sie pozycjami

funkcja(1, 2, 3, 4)


# poslugiwanie sie nazwami

funkcja(d=1, c=2, b=3, a=4)


# podejscie mieszane:


funkcja(1, b=20, c=30, d=40)
```

```python
def funkcja(a, b, *, c=1, d=2):  # snake_case
    print(a, b, c, d)

# wymuszamy to:
funkcja(1, 2)
funkcja(1, 2, c=3, d=4)

# nie pozwalany na to
# funkcja(1, 2, 3, 4)
```

```python
def funkcja(a, b, /,  *, c=1, d=2):  # snake_case
    print(a, b, c, d)


# wymuszamy to:
funkcja(1, 2, c=3, d=4)

# nie pozwalamy na to
# funkcja(a=1, b=2, c=3, d=4)
```

```python
def funkcja(a, b, /, *args, c=1, d=2, **kwargs):  # snake_case
    print(a, b, c, d)
    print("args", args)
    print("kwargs", kwargs)

funkcja(1, 2, 3, 4, c=3, d=4)
funkcja(1, 2, 3, 4, 5, 6, 7, 8, 9, c=3, d=4)
funkcja(1, 2, 3, 4, 5, 6, 7, 8, 9, c=3, d=4, e=600, g=300)
```
