```python
from functools import wraps



def log(func):

    @wraps(func)
    def wrapper(*args, **kwargs):
        print("jestem wrapperem")
        r = func(*args)
        
        return r
    # wrapper.__name__ = func.__name__
    # wrapper.__doc__ = func.__doc__
    # te powyzsze rzeczy robi wraps
    
    return wrapper

@log("x")
def foo():
    """Doc string funkcji foo"""
    print("cos tam")


#foo = log(foo)


@log
def bar(x):
    print("cos tam", x)

@log
def baz(x, y, *, z=1):
    return x + y + z
```
