# 🧠 **PROSTE wyjaśnienie: dlaczego różne wersje Pythona dają różne wyniki**

Wyobraź sobie, że masz **kilka osób**, które mają wspólnie liczyć licznik:

```
counter = 0
counter += 1
```

Każdy wątek to jedna osoba.

---

# 🔵 **Python 3.9 – stary system**

W Pythonie 3.9 działa **GIL** (Global Interpreter Lock).

To oznacza:

> „W jednym momencie tylko jedna osoba może wykonywać instrukcje Pythona”.

Czyli:

* Wątek A liczy → GIL blokuje innych
* Wątek B czeka
* Wątek C czeka

Dzięki temu:

* nikt nie wchodzi drugiemu w drogę,
* operacje są wykonywane *po kolei*,
* ale czasem Python może przerwać wątek w niefortunnym miejscu → pojawiają się drobne błędy.

Dlatego w 3.9:

* wynik jest **często poprawny**,
* ale czasem **zdarzy się błąd**.

To jak ludzie, którzy stoją w kolejce, ale czasem ktoś się przepchnie.

---

# 🟢 **Python 3.14 – ulepszony system (z GIL)**

Domyślny Python 3.14 **dalej ma GIL**.
Czyli nadal:

> Jedna osoba liczy naraz. Reszta czeka.

ALE:
Python 3.14 **usprawnił działanie wewnętrzne**:

* mniej przełączania między wątkami
* szybsze operacje
* mniej przypadków „przerwania w złym momencie”

Dlatego:

➡️ **wynik jest zawsze poprawny**,
➡️ nawet jeśli kod *teoretycznie* nie jest thread-safe.

To jak kolejka z ochroniarzem, który pilnuje, żeby nikt się nie przepychał.

---

# 🔴 **Python 3.14t – nowy system FREE-THREADED (bez GIL)**

To jest **inna wersja Pythona**.
Litera **t** oznacza *threading* → **brak GIL-a**.

Teraz:

> Wiele osób może jednocześnie ruszyć do liczenia.

I wtedy dzieją się rzeczy jak w realnym życiu:

* dwie osoby przeczytają tę samą wartość,
* obie ją zwiększą,
* obie zapiszą swój wynik,
* jedna nadpisze drugą,
* część operacji się zgubi.

Efekt:

➡️ **Wynik jest złym numerem**
➡️ za każdym uruchomieniem inny
➡️ czasem Python dostaje błędy pamięci („rozrzuciliśmy kartki z licznikiem po pokoju”)

To jakbyśmy powiedzieli ludziom:

> „Liczymy wszyscy naraz, bez żadnych zasad!”

No i wiadomo — chaos.


````python
%%writefile thr_example.py
import threading

counter = 0

def add():
    global counter
    for _ in range(100000):
        counter += 1

threads = [threading.Thread(target=add) for _ in range(2)]
[t.start() for t in threads]
[t.join() for t in threads]

print(counter)
````

````python
class X:
    def f(self): return "X"

class Y(X):
    pass

class Z(X):
    def f(self): return "Z"

class M:
    def f(self): return "M"

class A(Y, Z, M):
    pass

print(A().f())
````

- A. `X`
- B. `Y`
- C. `Z`
- D. `M`



**Poprawna: C**

> Wyjaśnienie: MRO klasy `A` odwiedza `Y`, potem `Z`, więc pierwsza znaleziona metoda `f` pochodzi z `Z`.

## **39. Co wypisze kod — asyncio task scheduling + event loop?**

```python
import asyncio

async def f():
    await asyncio.sleep(0.1)
    return 1

async def g():
    task = asyncio.create_task(f())
    await asyncio.sleep(0.2)
    return task.result()

```



````python
import asyncio

async def f():
    await asyncio.sleep(0)
    return 1

async def g():
    task = asyncio.create_task(f())
    await asyncio.sleep(0)
    return task.result()

print(asyncio.run(g()))
````

- A. `1`
- B. `None`
- C. `Task pending`
- D. `RuntimeError`



**Poprawna: A**

> Wyjaśnienie: Po `await asyncio.sleep(0.1)` zadanie `f` zdąży się wykonać i `task.result()` zwraca `1`.  A co będzie jak beda tam 0

## **40. Co zwróci kod — manualne tworzenie klasy przez `type()` i nadpisywanie `__call__`?**

```python
def make():
    def __call__(self, x):
        return x * 3
    return type("K", (), {"__call__": __call__})

K = make()
obj = K()
print(obj(5))
```



- A. `5`
- B. `15`
- C. `TypeError`
- D. `None`



**Poprawna: B**

> Wyjaśnienie: Klasa tworzona przez `type()` dostaje `__call__`, więc instancja zwraca `x * 3` jak funkcja.
