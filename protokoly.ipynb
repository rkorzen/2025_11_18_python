````python
class Countdown:
    def __init__(self, start):
        self.current = start

    def __iter__(self): return self

    def __next__(self):
        if self.current == 0:
            raise StopIteration
        value = self.current
        self.current -= 1
        return value
            


c = Countdown(3)
next(c) # 3
next(c) # 2
# next(c) # 1
# next(c) # StopIterarton

for c in Countdown(5):
    print(c)
        
````

````python
class EvenNumbers:
    def __init__(self, limit):
        self.limit = limit
        self.i = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.i == self.limit:
            raise StopIteration
        value = self.i * 2
        self.i += 1
        return value


for n in EvenNumbers(5):
    print(n)
# 0
# 2
# 4
# 6
# 8
````

### Context manager

with manager() <as f>:
   ... poczatek bloku kodu
   <blok kody>
... koniec 


```
__enter__
__exit__
```




````python
class MyManager:
    def __enter__(self):
        print("wejscioe do kontekstu")
        return "to bedzie w zmienne po slowie as"

    def __exit__(self, exc_type, exc, tb):
        print("wyjscie z kontekstu")
        print(exc_type)
````

````python
with MyManager() as m:
    print("W zmiennej m jest, ", m)
    1/0

print("Po context managerze")
````

````python
try:
    f = open("xxx", "w")
    1/0
finally:
    print("finally wykonuje sie zawsze")
    f.close()
````

````python
from contextlib import contextmanager
````

````python
@contextmanager
def my_manager():
    print("poczatek")
    yield "Wartosc"
    print("Koniec")

with my_manager() as f:
    print("xxx", "f")
````

````python
class SuppressValueError:
    def __enter__(self):
        return

    def __exit__(self, exc_type, exc, tb):
        if exc_type is ValueError:
            return True
        return False



with SuppressValueError():
    # int("x")
    1/0
    
print(10)
````

````python
debug = False

class EnableDebug:
    def __enter__(self):
        global debug
        debug = True

    def __exit__(self, exc_type, exc, tb):
        global debug
        debug = False
````

````python
if debug:
    print("A")

with EnableDebug():
    if debug:
        print("B")


if debug:
    print("C")
````

### Deksryptor

```
__get__
__set__
__delete__

__set_name__
```


````python


class Item:

    def __init__(self, price, length):
        self.price = price
        self.lenght = length


    @property
    def price(self):
        return self._price

    @price.setter
    def price(self, value):
        if value < 0: raise ValueError
        self._price = value

item = Item(1, 1)

vars(item), item.__dict__
````

````python
class Positive:

    def __get__(self, instance, owner):
        return instance.__dict__[self.name]

    def __set__(self, instance, value):
        if value < 0:
            raise ValueError
        instance.__dict__[self.name] = value

    def __set_name__(self, owner, name):
        self.name = name


class Item:
    price = Positive()
    length = Positive()

    def __init__(self, price, length):
        self.price = price
        self.lenght = length

item = Item(1, 1)
````
