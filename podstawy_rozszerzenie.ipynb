```python
lista = ["chleb", "ziemniaki", "pomarancza", "sdsd3dedfv345  r4tweger"]
ceny = [10, 20, 300.5, 1]
```

```python
paragon = ""

for i, el in enumerate(lista):
    cena = ceny[i]
    paragon += f"{el} - {cena} PLN\n"

print(paragon)
```

```python
paragon = ""

def string_len(x):
    return len(x)

po_dlugosci = sorted(lista, key=string_len)
najdluzszy = po_dlugosci[-1]
dlugosc = len(najdluzszy)


for i, el in enumerate(lista):
    cena = ceny[i]
    paragon += f"{el:{dlugosc}} - {cena:>3.2f} PLN\n"

print(paragon)
```

### for

```python
Mamy liste zawierajaca jakie liczby

dane = [1, 10, 30, 31, 12, 56, 3, 5, 7, 11]
l_pierwsze = []

...

assert l_pierwsze == [31, 3, 5, 7, 11]

napisz kod ktory wybierze z tej listy liczby pierwsze. Nie posluguj sie funkcjami (wlasnymi)
```

```python
dane = [1, 10, 30, 31, 12, 56, 3, 5, 7, 11, 2]
l_pierwsze = []


for liczba in dane:
    if liczba < 2:
        continue
    for dzielnik in range(2, liczba):
        if liczba % dzielnik == 0:
            break
    else:
        l_pierwsze.append(liczba)
        

assert l_pierwsze == [31, 3, 5, 7, 11, 2]
```
